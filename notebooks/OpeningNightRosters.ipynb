**FIRST THINGS FIRST**

```python
import pandas as pd
```

**Reading the 2026-27 CSV file into a DataFrame**

```python
df_2627 = pd.read_csv("C:/Users/g32ma/dataProjects/opening night rosters/data/Opening night rosters - 202627.csv")
```

**Reading the 2025-26 CSV file into a DataFrame**

```python
df_2526 = pd.read_csv("C:/Users/g32ma/dataProjects/opening night rosters/data/Opening night rosters - 202526.csv")
```

**Tagging every row in df_2526 with its season label so the source is clear after merging**

```python
df_2526['Season'] = '2025-26'
```

**Confirming the column was added**

```python
df_2526.head(3)
```

**Tagging every row in df_2627 with its season label so the source is clear after merging**

```python
df_2627['Season'] = '2026-27'
```

**Confirming the column was added**

```python
df_2627.head(3)
```

**Stacking both season tables together, using ignore_index=True to reset the row index and avoid duplicate numbers**

```python
df_master = pd.concat([df_2526, df_2627], ignore_index=True)
```

**Confirming by checking random sample**

```python
df_master.sample(5)
```

**Create a basic pivot table counting players per country for each season**

```python
country_pivot = df_master.pivot_table(index='Country', columns='Season', values='Name', aggfunc='count', fill_value=0)
```

**Confirming our pivot table was created**

```python
country_pivot.head()
```

**Sorting the table to see the top countries in the 2026-27 season**

```python
country_pivot = country_pivot.sort_values(by='2026-27', ascending=False)
country_pivot.head()
```

**Calculating absolute change from 2025-26 to 2026-27**

```python
country_pivot['Absolute_Change'] = (
    country_pivot['2026-27'] - country_pivot['2025-26']
)
country_pivot.head()
```

**Sorting by top change**

```python
country_pivot = country_pivot.sort_values(by='Absolute_Change', ascending=False)
country_pivot.head()
```

**Finding largest drop in absolute change**

```python
country_pivot = country_pivot.sort_values(by='Absolute_Change', ascending=True)
country_pivot.head()
```

**Calculating the year-over-year percentage change for each country**

```python
country_pivot['Pct_Change'] = (country_pivot['Absolute_Change'] / country_pivot['2025-26']) * 100

country_pivot.head()
```

**Sorting by top change**

```python
country_pivot = country_pivot.sort_values(by='Pct_Change', ascending=False)
country_pivot.head(15)
```

**And largest decline**

```python
country_pivot = country_pivot.sort_values(by='Pct_Change', ascending=True)
country_pivot.head(15)
```
